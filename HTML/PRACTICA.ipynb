```python
from pydoc import text


df_csv = pd.read_csv('ventas_tienda_festival.csv')
df_csv['importe'] = df_csv['precio_eur'] * df_csv['unidades']
filtrado = df_csv[df_csv['importe'] >= 30].copy()
# filtrado.to_html('mi nuevo archivo.html', index=False)

mi_tabla = filtrado.to_html(index=False)

croquis_html = f"""

<!DOCTYPE html>
<html>
<head>
    <meta charset="UTF-8">
    <title>Ventas filtradas</title>
</head>
    
<body>
<div style="background-color: #f0f0f0; padding: 10px; color: blue;">
    <h1>    
    Ventas_filtradas(importe ≥ 30 €)
    </h1>
    <p style="color: red;">Precios inigualables.</p>
</div>    
    
{mi_tabla}
</body>

</html>
"""

with open("mi_archivo.html", "w", encoding="utf-8") as f:
    f.write(croquis_html)


display(filtrado)
```
